- Lấy 3 file excel về nhưng chưa làm sạch: 

```python
import pandas as pd
import numpy as np
```

```python
# Lấy file campaign_info.csv: 
df_campaign = pd.read_csv('campaign_info.csv')
```

```python
# Lấy file customer_response.csv: 
df_customer_reponse = pd.read_csv('customer_response.csv')
```

```python
# Lấy file sales_conversion.csv: 
df_sales_conversion = pd.read_csv('sales_conversion.csv')
```

- Làm sạch 3 file : 


```python
# Chuẩn hóa file customer_response.csv : 

for cot in ["customer_id", "campaign_id"]:
    df_customer_reponse[cot] = df_customer_reponse[cot].astype(str).str.strip() 
    df_customer_reponse[cot] = df_customer_reponse[cot].str.title()
    

df_customer_reponse['response'] = df_customer_reponse['response'].astype(str).str.strip()
df_customer_reponse['response'] = df_customer_reponse['response'].str.replace(r"\s+", "", regex=True)
df_customer_reponse['response'] = df_customer_reponse['response'].str.title()
```

```python
# Chuẩn hóa file campaign_info.csv: 

# chuẩn hóa cột campaign_id
df_campaign['campaign_id'] = df_campaign['campaign_id'].astype(str).str.strip()
df_campaign['campaign_id'] = df_campaign['campaign_id'].str.upper()

# Chuẩn hóa cột channel: 
df_campaign['channel'] = df_campaign['channel'].astype(str).str.strip()
df_campaign['channel'] = df_campaign['channel'].str.replace(r'\s+','',regex=True)
df_campaign["channel"] = df_campaign["channel"].str.lower()

mask = {
    "facebok": "Facebook",
    "facebook": "Facebook",
    "google": "Google",
    "zalo": "Zalo",
    "tiktok": "TikTok",
    "instagram": "Instagram",
    'email': 'Email'

}

df_campaign['channel'] = df_campaign['channel'].map(mask)

# Chuẩn hóa cột budget: 
df_campaign['budget'] = df_campaign['budget'].astype(str).str.strip()
df_campaign['budget'] = df_campaign['budget'].str.replace(r'\s+','',regex=True)
df_campaign['budget'] = df_campaign['budget'].str.replace('tr','000000',regex=False)
df_campaign['budget'] = df_campaign['budget'].str.replace('.','',regex=False)
df_campaign['budget'] = df_campaign['budget'].str.replace(r'[^\d]','',regex=True)
df_campaign['budget'] = df_campaign['budget'].str.replace('Đ','',regex=False)


df_campaign['budget'] = pd.to_numeric(df_campaign['budget'], errors='coerce')


```

```python
# Chuẩn hóa sales_conversion.csv

# Chuẩn hóa cột customer_id: 
df_sales_conversion['customer_id'] = df_sales_conversion['customer_id'].astype(str).str.strip()
df_sales_conversion['customer_id'] = df_sales_conversion['customer_id'].str.replace(r'\s+','',regex=True)
df_sales_conversion['customer_id'] = df_sales_conversion['customer_id'].str.upper()

# Chuẩn hóa cột campaign_id: 
df_sales_conversion['campaign_id'] = df_sales_conversion['campaign_id'].astype(str).str.strip()
df_sales_conversion['campaign_id'] = df_sales_conversion['campaign_id'].str.replace(r'\s+','',regex=True)
df_sales_conversion['campaign_id'] = df_sales_conversion['campaign_id'].str.upper()


# Chuẩn hóa cột converted: 
df_sales_conversion['converted'] = df_sales_conversion['converted'].astype(str).str.strip()
df_sales_conversion['converted'] = df_sales_conversion['converted'].str.replace(r'\s+','',regex=True)
df_sales_conversion['converted'] = df_sales_conversion['converted'].str.replace('yes','1',regex=False)
df_sales_conversion['converted'] = pd.to_numeric(df_sales_conversion['converted'],errors='coerce')

# Chuẩn hóa cột amount: 
df_sales_conversion['amount'] = df_sales_conversion['amount'].astype(str).str.strip()
df_sales_conversion['amount'] = df_sales_conversion['amount'].str.replace(r'\s+','',regex=True)
df_sales_conversion['amount'] = df_sales_conversion['amount'].str.replace('k','000',regex=False)

df_sales_conversion['amount'] = pd.to_numeric(df_sales_conversion['amount'],errors='coerce')
```
