```python
import pandas as pd
df_campaign = pd.read_csv('../data_clean/campaign_clean_info.csv')
df_response = pd.read_csv('../data_clean/customer_response_clean.csv')
df_sales = pd.read_csv('../data_clean/sales_conversion_clean.csv')
```

1. Thống kê số khách hàng phản hồi theo từng chiến dịch

```python
print("--- BẮT ĐẦU NHIỆM VỤ 2 ---")
print("\n1. Số lượng phản hồi theo từng chiến dịch:")
response_counts = df_response['campaign_id'].value_counts()
print(response_counts)
top_5_campaigns = response_counts.iloc[:5]
print("\nTop 5 chiến dịch có phản hồi cao nhất:")
print(top_5_campaigns)
```

2. Tìm các chiến dịch không có phản hồi

```python
mask_no_response = ~df_campaign['campaign_id'].isin(df_response['campaign_id'])
campaigns_no_response = df_campaign.loc[mask_no_response]

print(f"\n2. Các chiến dịch không có phản hồi:")
if campaigns_no_response.empty:
    print("Tất cả chiến dịch đều có phản hồi.")
else:
    print(campaigns_no_response[['campaign_id', 'channel']])
```

3. Thống kê tỷ lệ Click / View / Ignore

```python
print("\n3. Tỷ lệ các loại phản hồi:")
response_rates = df_response['response'].value_counts(normalize=True) * 100
print(response_rates)
clicks_only = df_response.loc[df_response['response'] == 'Click']
print(f"Tổng số lượt Click chi tiết: {len(clicks_only)}")
```

4. Tính chuyển đổi trung bình của từng chiến dịch

```python
avg_conversion = df_sales.groupby('campaign_id')['converted'].mean()
print("\n4. Tỷ lệ chuyển đổi trung bình theo từng chiến dịch:")
print(avg_conversion)
```

5. Tìm khách hàng có phản hồi nhưng không có dữ liệu chuyển đổi

```python
customers_with_sales_data = df_sales['customer_id'].unique()
mask_missing_conversion = ~df_response['customer_id'].isin(customers_with_sales_data)
customers_no_conversion_data = df_response.loc[mask_missing_conversion]

print(f"\n5. Số lượng khách phản hồi nhưng thiếu dữ liệu chuyển đổi: {len(customers_no_conversion_data)}")
print("Danh sách 10 dòng đầu tiên")
print(customers_no_conversion_data.iloc[:10])
```
