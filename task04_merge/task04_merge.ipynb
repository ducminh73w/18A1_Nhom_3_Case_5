1. Đọc 3 file dữ liệu sạch từ thư mục data_clean/


```python
import pandas as pd

campaign = pd.read_csv("../data_clean/campaign_clean_info.csv")
response = pd.read_csv("../data_clean/customer_response_clean.csv")
sales = pd.read_csv("../data_clean/sales_conversion_clean.csv")


print(" Đọc dữ liệu hoàn tất ")
print("campaign:", campaign.shape)
print("response:", response.shape)
print("sales:", sales.shape)
```

2. Merge campaign_info với customer_response theo campaign_id

```python
df_merge1 = pd.merge(
    campaign,
    response,
    on="campaign_id",
    how="left"
)

print("Merge campaign_info với customer_response theo campaign_id hoàn tất")
print(df_merge1.shape)
```

3. Merge thêm sales_conversion theo customer_id

```python
df_full = pd.merge(
    df_merge1,
    sales,
    on="customer_id",
    how="left"
)

print(" Merge sales_conversion theo customer_id hoàn tất")
print(df_full.shape)
```

4. Phát hiện mismatch giữa các file

```python
mismatch_campaign = response[
    ~response["campaign_id"].isin(campaign["campaign_id"])
]

mismatch_customer = sales[
    ~sales["customer_id"].isin(response["customer_id"])
]

print(" Phát hiện mismatch:")
print("Mismatch campaign_id:", mismatch_campaign.shape)
print("Mismatch customer_id:", mismatch_customer.shape)
```

5. Xuất kết quả vào thư mục task04_merge/

```python
df_full.to_csv("merged_full_data.csv", index=False)
mismatch_campaign.to_csv("mismatch_campaign.csv", index=False)
mismatch_customer.to_csv("mismatch_customer.csv", index=False)

print(" Xuất file hoàn tất")
print("Đã tạo merged_full_data.csv")
print("Đã tạo mismatch_campaign.csv")
print("Đã tạo mismatch_customer.csv")
```
