1.Pivot tỷ lệ chuyển đổi theo kênh × loại phản hồi

```python
import pandas as pd

df = pd.read_csv('../task04_merge/merged_full_data.csv')

df.head()
conversion_summary = df.groupby(['channel', 'response']).agg(
    Total_Responses=('campaign_id_x', 'size'),
    Total_Conversions=('converted', 'sum')
).reset_index()

conversion_summary['Conversion_Rate'] = (
    conversion_summary['Total_Conversions'] / conversion_summary['Total_Responses']) * 100
pivot_conversion_rate = conversion_summary.pivot_table(
    index='channel',
    columns='response',
    values='Conversion_Rate',
    fill_value=0,
    margins=False,                
    dropna=True,                  
    margins_name='All',         
    observed=False               
).applymap(lambda x: f"{x:.2f}%")

print("\n--- PIVOT: TỶ LỆ CHUYỂN ĐỔI ---")
print(pivot_conversion_rate)
df.head(20)
```

2.Pivot doanh thu theo chiến dịch × kênh

```python
pivot_revenue = pd.pivot_table(df, 
                               values='amount', 
                               index='campaign_id_x', 
                               columns='channel', 
                               aggfunc='mean', 
                               fill_value=0)
print("\n=== PIVOT DOANH THU THEO CHIẾN DỊCH × KÊNH ===")
print(pivot_revenue)
```

3. Chuyển đổi stack/unstack để hiểu MultiIndex

```python
# Stack: 
stacked = pivot_conversion_rate.stack()
print("\nDữ liệu sau khi stack (MultiIndex):\n", stacked)
# Unstack: 
unstacked = stacked.unstack()
print("\nDữ liệu sau khi unstack (quay về pivot):\n", unstacked)
```

4. Nhận xét chiến dịch/kênh hiệu quả nhất

```python
if 'Click' in pivot_conversion_rate.columns:
    best_channel = pivot_conversion_rate['Click'].str.rstrip('%').astype(float).idxmax()
    print("Kênh hiệu quả nhất:", best_channel)
best_campaign = pivot_revenue.mean(axis=1).idxmax()
print("Chiến dịch hiệu quả nhất:", best_campaign)
```
