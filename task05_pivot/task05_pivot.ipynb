```python
import pandas as pd

campaign = pd.read_csv("../data_clean/campaign_clean_info.csv")
response = pd.read_csv("../data_clean/customer_response_clean.csv")
sales = pd.read_csv("../data_clean/sales_conversion_clean.csv")


print(" Đọc dữ liệu hoàn tất ")
print("campaign:", campaign.shape)
print("response:", response.shape)
print("sales:", sales.shape)
```

1.Pivot tỷ lệ chuyển đổi theo kênh × loại phản hồi

```python
import pandas as pd
merged_full_data=pd.read_csv("merged_full_data.csv")
pivot_conversion = pd.pivot_table(merged_full_data,
                                values='converted',
                                index='channel',
                                columns='response',
                                aggfunc='mean', 
                                fill_value=0
)
print("Pivot tỷ lệ chuyển đổi theo kênh × loại phản hồi:\n", pivot_conversion)
```

2.Pivot doanh thu theo chiến dịch × kênh

```python
pivot_revenue = pd.pivot_table(merged_full_data, 
                               values='amount', 
                               index='campaign_id', 
                               columns='channel', 
                               aggfunc='mean', 
                               fill_value=0
)
print("\nPivot doanh thu theo chiến dịch × kênh:\n", pivot_revenue)
```

3. Chuyển đổi stack/unstack để hiểu MultiIndex

```python
# Stack: Chuyển columns thành rows (tạo MultiIndex)
stacked = pivot_conversion.stack()
print("\nDữ liệu sau khi stack (MultiIndex):\n", stacked)
# Unstack: Chuyển lại từ rows thành columns
unstacked = stacked.unstack()
print("\nDữ liệu sau khi unstack (quay về pivot):\n", unstacked)
```

4. Nhận xét chiến dịch/kênh hiệu quả nhất

```python
# Dựa trên pivot_conversion: Kênh có tỷ lệ chuyển đổi cao nhất cho 'Click' 
best_channel = pivot_conversion['Click'].idxmax()  # Giả định hiệu quả dựa trên Click
best_campaign = pivot_revenue.mean(axis=1).idxmax()  # Chiến dịch có doanh thu trung bình cao nhất
print(f"\nNhận xét: Kênh hiệu quả nhất (dựa trên tỷ lệ chuyển đổi Click): {best_channel}")
print(f"Chiến dịch hiệu quả nhất (dựa trên doanh thu trung bình): {best_campaign}")
```
